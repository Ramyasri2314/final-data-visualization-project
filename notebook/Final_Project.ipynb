# Evolution of Global Space Exploration: A Historical Analysis of Space Missions (1957–2022)

## Introduction

Space exploration has played a significant role in advancing science, technology, and human understanding of the universe since the launch of the first artificial satellite in 1957. Over the decades, space missions have evolved from government-led programs to commercial space ventures involving organizations such as NASA, SpaceX, ISRO, ESA, and Roscosmos.

This project analyses historical global space mission data from 1957 to 2022 to explore trends in launch activity, mission success, launch organizations, and launch locations. Using interactive Plotly visualizations and a Streamlit dashboard, the project aims to uncover meaningful insights into the evolution of global space exploration through data-driven analysis.

## Dataset Description

This project uses the **Space Missions** dataset, which contains historical records of global space missions conducted between **1957 and 2022**. The dataset includes information about launch companies, launch locations, launch dates and times, rocket names, mission names, rocket status, mission status, and launch prices (where available).

The dataset consists of **4,630 records** and **9 columns**, making it suitable for analysing long-term trends in global space exploration. It provides temporal, geographical, and categorical information that supports meaningful analytical questions and interactive data visualisation.

```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots
```

```python
df = pd.read_csv("space_missions.csv", encoding="latin1")

df.head()
```

```python
df = df.drop_duplicates()

df.shape
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Date"].head()
```

```python
df["Year"] = df["Date"].dt.year

df[["Date", "Year"]].head()
```

```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots
```

```python
df["Month"] = df["Date"].dt.month_name()

df[["Date", "Month"]].head()
```

## Data Cleaning

Data cleaning is an important step before analysis because it improves the quality and reliability of the dataset. In this project, duplicate records were removed, date values were converted into a proper datetime format, and additional time-based features were created to support trend analysis.

```python
# Number of space missions launched each year
missions_per_year = (
    df.groupby("Year")
      .size()
      .reset_index(name="Number of Missions")
)

missions_per_year.head()
```

```python
fig = px.line(
    missions_per_year,
    x="Year",
    y="Number of Missions",
    title="Growth of Global Space Missions Over Time (1957–2022)",
    markers=True
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Year",
    yaxis_title="Number of Missions"
)

fig.show()
```

### Interpretation

The visualization illustrates the historical growth of global space missions from 1957 to 2022. During the early years of space exploration, the number of launches was relatively low because only a few countries had the capability to conduct space missions. Over time, launch activity increased significantly due to technological advancements, international collaboration, and the emergence of private aerospace companies. The overall upward trend demonstrates the continuous expansion and growing importance of global space exploration.

## Analytical Question 2

### Which space organizations have conducted the highest number of space missions?

Space agencies and private aerospace companies have played a significant role in the advancement of global space exploration. Comparing the number of missions conducted by different organizations helps identify the most active contributors to space exploration and highlights the transition from government-led programs to commercial spaceflight.

```python
company_missions = (
    df["Company"]
    .value_counts()
    .head(10)
    .reset_index()
)

company_missions.columns = ["Company", "Number of Missions"]

company_missions
```

```python
fig = px.bar(
    company_missions,
    x="Number of Missions",
    y="Company",
    orientation="h",
    title="Top 10 Space Organizations by Number of Missions",
    text="Number of Missions",
    color="Number of Missions"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Number of Missions",
    yaxis_title="Space Organization",
    yaxis=dict(categoryorder="total ascending")
)

fig.show()
```

### Interpretation

The chart highlights the ten organizations that have conducted the highest number of space missions. Government space agencies dominate the historical launch records because they pioneered space exploration during the early decades of the space race. In recent years, commercial companies have become increasingly active, reflecting the rapid growth of the private space industry and its contribution to modern space exploration.

## Analytical Question 3

### How have mission success rates changed over time?

Mission success is one of the most important indicators of technological progress in space exploration. Analysing mission outcomes over different years helps identify improvements in launch reliability and the evolution of space technology across decades.

```python
mission_success = (
    df.groupby(["Year", "MissionStatus"])
      .size()
      .reset_index(name="Count")
)

mission_success.head()
```

```python
fig = px.line(
    mission_success,
    x="Year",
    y="Count",
    color="MissionStatus",
    markers=True,
    title="Mission Success and Failure Trends Over Time"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Year",
    yaxis_title="Number of Missions"
)

fig.show()
```

### Interpretation

The visualization compares successful and unsuccessful space missions over time. Successful missions consistently outnumber failed missions, and the gap widens in recent decades. This trend suggests significant improvements in rocket engineering, mission planning, and operational reliability, reflecting the technological advancement of the global space industry.

## Analytical Question 4

### Which launch locations have hosted the highest number of space missions?

Launch locations play a crucial role in space exploration because geographical position, infrastructure, and environmental conditions influence mission planning and launch success. This analysis identifies the most active launch sites and compares their contribution to global space exploration.

```python
top_locations = (
    df.groupby("Location")
      .size()
      .reset_index(name="Number of Missions")
      .sort_values(by="Number of Missions", ascending=False)
      .head(10)
)

top_locations
```

```python
fig = px.bar(
    top_locations,
    x="Number of Missions",
    y="Location",
    orientation="h",
    color="Number of Missions",
    text="Number of Missions",
    title="Top 10 Launch Locations by Number of Space Missions"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Number of Missions",
    yaxis_title="Launch Location",
    yaxis=dict(categoryorder="total ascending"),
    coloraxis_showscale=False
)

fig.show()
```

### Interpretation

The analysis reveals that a small number of launch sites account for a large proportion of global space missions. Facilities such as Baikonur Cosmodrome, Cape Canaveral, and other major launch centres dominate because they possess advanced infrastructure, strategic geographical advantages, and a long history of supporting both government and commercial space programs. These launch sites continue to play a significant role in the global space industry.

## Analytical Question 5

### Which space organizations have achieved the highest mission success rates?

Mission success rate is an important indicator of the reliability and operational performance of space organizations. Comparing the success rates of different organizations helps evaluate their effectiveness and technological capabilities in conducting successful space missions.

```python
company_success = (
    df.groupby(["Company", "MissionStatus"])
      .size()
      .unstack(fill_value=0)
)

company_success["Total Missions"] = company_success.sum(axis=1)
company_success["Success Rate (%)"] = (
    company_success["Success"] / company_success["Total Missions"]
) * 100

top_success = (
    company_success[company_success["Total Missions"] >= 20]
    .sort_values("Success Rate (%)", ascending=False)
    .head(10)
    .reset_index()
)

top_success
```

```python
fig = px.bar(
    top_success,
    x="Success Rate (%)",
    y="Company",
    orientation="h",
    color="Success Rate (%)",
    text="Success Rate (%)",
    title="Top 10 Space Organizations by Mission Success Rate"
)

fig.update_traces(texttemplate="%{text:.1f}%", textposition="outside")

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Mission Success Rate (%)",
    yaxis_title="Space Organization",
    yaxis=dict(categoryorder="total ascending"),
    coloraxis_showscale=False
)

fig.show()
```

### Interpretation

The analysis shows that several space organizations have achieved exceptionally high mission success rates. A consistently high success rate indicates strong engineering capabilities, effective mission planning, and operational excellence. Comparing organizations based on success rate provides a more meaningful assessment of performance than comparing only the total number of missions.

## Analytical Question 6

### How has mission activity evolved across different decades?

Analysing space missions by decade provides a broader understanding of the evolution of global space exploration. This analysis highlights major periods of growth, technological advancement, and increased participation in space activities over time.

```python
df["Decade"] = (df["Year"] // 10) * 10

missions_decade = (
    df.groupby("Decade")
      .size()
      .reset_index(name="Number of Missions")
)

missions_decade
```

```python
fig = px.bar(
    missions_decade,
    x="Decade",
    y="Number of Missions",
    color="Number of Missions",
    text="Number of Missions",
    title="Growth of Global Space Missions by Decade"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Decade",
    yaxis_title="Number of Missions",
    coloraxis_showscale=False
)

fig.show()
```

### Interpretation

The number of global space missions has increased considerably across successive decades. While the early decades were dominated by government-led space programmes, later decades experienced rapid growth due to technological innovation, international collaboration, and the expansion of commercial space companies. The results illustrate the continuous evolution of the global space industry.

## Analytical Question 7

### Which rocket types have been used most frequently in global space missions?

Rocket technology has evolved significantly over the history of space exploration. Identifying the most frequently used rocket types provides insights into the launch vehicles that have played a major role in supporting scientific research, satellite deployment, and commercial space missions.

```python
fig = px.bar(
    top_rockets,
    x="Number of Missions",
    y="Rocket",
    orientation="h",
    color="Number of Missions",
    text="Number of Missions",
    title="Top 10 Most Frequently Used Rockets"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Number of Missions",
    yaxis_title="Rocket",
    yaxis=dict(categoryorder="total ascending"),
    coloraxis_showscale=False
)

fig.show()
```

### Interpretation

The analysis identifies the rocket types that have been used most frequently throughout the history of space exploration. These launch vehicles have supported a large number of successful missions because of their reliability, operational efficiency, and long-term deployment. ## Analytical Question 8

### Are there seasonal patterns in global space missions?

The timing of space launches may be influenced by weather conditions, mission requirements, launch schedules, and operational planning. This analysis investigates whether certain months experience a higher number of space missions than others.

```python
missions_month = (
    df.groupby("Month")
      .size()
      .reset_index(name="Number of Missions")
)

# Arrange months in calendar order
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

missions_month["Month"] = pd.Categorical(
    missions_month["Month"],
    categories=month_order,
    ordered=True
)

missions_month = missions_month.sort_values("Month")

missions_month
```

```python
fig = px.line(
    missions_month,
    x="Month",
    y="Number of Missions",
    markers=True,
    title="Monthly Distribution of Global Space Missions"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Month",
    yaxis_title="Number of Missions"
)

fig.show()
```

### Interpretation

The monthly distribution of space missions indicates whether launch activity is evenly spread throughout the year or concentrated in specific months. Any noticeable peaks may reflect favourable weather conditions, operational schedules, or mission planning strategies adopted by different space organisations.

## Analytical Question 9

### What is the distribution of active and retired rockets used in global space missions?

Rocket status reflects the evolution of launch vehicle technology over time. Analysing the proportion of active and retired rockets provides insights into technological advancements, fleet modernisation, and the transition from older launch systems to newer and more efficient rocket designs.

```python
rocket_status = (
    df["RocketStatus"]
    .value_counts()
    .reset_index()
)

rocket_status.columns = ["Rocket Status", "Count"]

rocket_status
```

```python
fig = px.pie(
    rocket_status,
    names="Rocket Status",
    values="Count",
    hole=0.5,
    title="Distribution of Active and Retired Rockets"
)

fig.update_traces(
    textposition="inside",
    textinfo="percent+label"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

### Interpretation

The donut chart illustrates the proportion of active and retired rockets used in global space missions. A higher proportion of retired rockets reflects the long history of space exploration, where older launch vehicles have been replaced by more advanced technologies. The presence of active rockets highlights the continuous innovation and ongoing development of modern launch systems.

## Analytical Question 10

### How do mission success rates compare among the top space organizations?

The success rate of space missions is a key indicator of an organization's technological capability and operational reliability. This analysis compares the mission success rates of the leading space organizations with the highest number of missions, providing insights into their performance and consistency over time.

```python
# Select top 10 organizations by number of missions
top_companies = df["Company"].value_counts().head(10).index

# Filter data
company_status = df[df["Company"].isin(top_companies)]

# Calculate success rate
success_rate = (
    company_status.groupby(["Company", "MissionStatus"])
    .size()
    .unstack(fill_value=0)
)

success_rate["Total Missions"] = success_rate.sum(axis=1)
success_rate["Success Rate (%)"] = (
    success_rate["Success"] / success_rate["Total Missions"]
) * 100

success_rate = (
    success_rate.sort_values("Success Rate (%)", ascending=False)
    .reset_index()
)

success_rate
```

```python
fig = px.bar(
    success_rate,
    x="Company",
    y="Success Rate (%)",
    color="Success Rate (%)",
    text="Success Rate (%)",
    title="Mission Success Rate of Top Space Organizations"
)

fig.update_traces(texttemplate="%{text:.1f}%", textposition="outside")

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Space Organization",
    yaxis_title="Mission Success Rate (%)",
    xaxis_tickangle=-45,
    coloraxis_showscale=False
)

fig.show()
```

### Interpretation

The comparison reveals noticeable differences in mission success rates among the leading space organizations. While several organizations have maintained consistently high success rates, others have experienced comparatively lower performance due to technological challenges, experimental missions, or historical limitations. Overall, the analysis highlights continuous improvements in mission reliability and demonstrates the advancement of modern space exploration.
